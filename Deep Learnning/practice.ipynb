```python
import torch  # import the PyTorch library for tensor operations
```

```python
x = torch.randint(0, 20, (2, 3, 4))  # create a random integer tensor with shape [2,3,4]
```

```python
x  # display the tensor values
```

```python
x.shape  # show tensor shape: 2 blocks, 3 rows, 4 columns
```

```python
x.dtype  # display the element data type of the tensor
```

```python
x.reshape(6, 4)  # reshape the tensor into shape [6,4]
```

```python
x.unsqueeze(0).shape  # add a new leading dimension and show the new shape
```

```python
y = x.flatten()  # flatten the tensor into a 1D vector
```

```python
y.shape  # show the shape of the flattened tensor
```

```python
z = torch.randint(0, 20, (2, 3, 4))  # create another random tensor with same shape
```

```python
z  # display the second tensor values
```

```python
conct = torch.concat([x, z])  # concatenate x and z along the first dimension
conct  # display the concatenated tensor
```

```python
stack_meth = torch.stack([x, z])  # stack x and z into a new dimension
stack_meth  # display the stacked tensor
```

```python
x.max()  # compute the maximum value in tensor x
```

```python
x.argmax()  # compute the index of the maximum value in the flattened tensor
```

```python
conct.sum()  # compute the sum of all elements in the concatenated tensor
```

```python
torch.sum(x)  # compute the sum of all elements in tensor x using torch.sum
```

```python
torch.relu(x)  # apply rectified linear unit activation to x
```
