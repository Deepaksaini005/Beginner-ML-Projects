```python
# cross product
a = pt.tensor([[1, 3], [2, 4]])
b = pt.tensor([[5, 7], [6, 8]])

print(pt.matmul(a, b))    # matmul is use for  cross product
```

```python
a @ b #second method for cross product
```

```python
a = pt.tensor([1, 3, 2])
b = pt.tensor([5, 7, 6])

result = pt.dot(a, b)
print(result)
```
