```python
# Import TensorFlow library
import tensorflow as tf
```

```python
# Create a 2D tensor with values 1 to 6
x = tf.constant([[1, 2, 3], [4, 5, 6]])

# Print the tensor
print(x)
```

```python
# Import NumPy for array handling
import numpy as np

# Create a NumPy array
x = np.array([1, 2, 3])

# Convert NumPy array into a TensorFlow tensor
tensor = tf.convert_to_tensor(x)

# Check the data type of the tensor
tensor.dtype
```

```python
# Convert the tensor to float32 data type
x = tf.cast(x, tf.float32)
```

```python
# Check the current data type of x
x.dtype
```

```python
# Generate a random tensor of shape (4, 4)
tf.random.normal((4, 4))
```

```python
# Generate a random tensor with uniform values and a minimum value of 10
tf.random.uniform((3, 3), minval=10)
```

```python
# Generate a random tensor with uniform values between 10 and 20
tf.random.uniform((3, 3), minval=10, maxval=20)
```
