```python
# Import TensorFlow library
import tensorflow as tf
```

```python
# Create a 2D tensor with values 1 to 6
x = tf.constant([[1, 2, 3], [4, 5, 6]])

# Print the tensor
print(x)
```

```python
# Import NumPy for array handling
import numpy as np

# Create a NumPy array
x = np.array([1, 2, 3])

# Convert NumPy array into a TensorFlow tensor
tensor = tf.convert_to_tensor(x)

# Check the data type of the tensor
tensor.dtype
```

```python
# Convert the tensor to float32 data type
x = tf.cast(x, tf.float32)
```

```python
# Check the current data type of x
x.dtype
```

```python
# Generate a random tensor of shape (4, 4)
tf.random.normal((4, 4))
```

```python
# Generate a random tensor with uniform values and a minimum value of 10
tf.random.uniform((3, 3), minval=10)
```

```python
# Generate a random tensor with uniform values between 10 and 20
tf.random.uniform((3, 3), minval=10, maxval=20)
```

```python
# Create a 1D tensor with values 1, 2, 3
x = tf.constant([1, 2, 3])
# Display the tensor
x
```

```python
# Create a TensorFlow variable with two values
y = tf.Variable([0.5, 1])
```

```python
# Convert x tensor to float16 data type
x = tf.cast(x, tf.float16)
```

```python
# Reshape x from shape (3,) to shape (3, 1)
x = tf.reshape(x, (3, 1))
# Display the reshaped tensor
x
```

```python
# Remove size-one dimensions from x
tf.squeeze(x)
```

```python
# Add a new dimension at axis 0
tf.expand_dims(x, axis=0)
```

```python
# Create a 1D tensor with four values
z = tf.constant([1, 2, 3, 4])
# Display the tensor
z
```

```python
# Reshape z into a 2x2 matrix
z = tf.reshape(z, (2, 2))
# Display the reshaped tensor
z
```

```python
# Remove any size-one dimensions from z
tf.squeeze(z)
```

```python
# Create a 3x4 tensor filled with zeros
a = tf.zeros((3, 4))
# Display the zero tensor
a
```

```python
# Create a 2x5 tensor filled with ones
one = tf.ones((2, 5))
# Display the ones tensor
one
```

```python
# Create a 3x3 tensor filled with the value 7
s = tf.fill((3, 3), 7)
# Display the filled tensor
s
```

```python
# Create a tensor with values 0 through 9
new = tf.range(10)
# Display the range tensor
new
```

```python
# Create a tensor with values 10,20,...,100
old = tf.range(10, 110, 10)
# Display the range tensor
old
```

```python
# Join the new and old tensors along axis 0
tf.concat([new, old], axis=0)
```

```python
# Compute the mean of new
tf.reduce_mean(new)
```

```python
# Compute the sum of new
tf.reduce_sum(new)
```

```python
# Shuffle the elements of old randomly
tf.random.shuffle(old)
```
