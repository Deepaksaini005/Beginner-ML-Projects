Task 1 – Horizontal Concatenation (Easy)
Create two tensors:

A = [[1,2], [3,4]]

B = [[5,6], [7,8]]

Requirements
Concatenate them using axis=1.
Print:
Original shapes ,
Final shape ,
Resulting tensor

```python

A = tf.constant([[1, 2], [3, 4]])
B = tf.constant([[5, 6], [7, 8]])
```

```python

C = tf.concat([A, B], axis=1)
C
```

```python

print("A shape:", A.shape)
print("B shape:", B.shape)
print("C shape:", C.shape)
```
