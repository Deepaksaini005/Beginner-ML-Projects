```python
import os  # for file and directory operations
import numpy as np  # numerical computing; not used here yet but often helpful
import pandas as pd  # data manipulation library; can load labels or metadata if needed
import matplotlib.pyplot as plt  # visualization library for plotting images or graphs

import torch  # PyTorch core library for building models and tensors
from torch.utils.data import Dataset  # base class to create a custom dataset

from PIL import Image  # image loading and basic image processing
```

```python
class CustomDataset(Dataset):
    def __init__(self, dir):
        # store the path to the image folder
        self.dir = dir
        # list all files in the folder (images expected)
        self.images = os.listdir(dir)

    def __len__(self):
        # return the total number of images in the dataset
        return len(self.images)

    def __getitem__(self, index):
        # build the full path to the selected image file
        image_path = os.path.join(self.dir, self.images[index])
        # open the image using PIL and return it
        image = Image.open(image_path)

        return image
```

```python
# path to the folder containing horse images
# update this path if your dataset is stored elsewhere
data_path = "C:/Users/saini/Downloads/archive (2)/data/horses"
```

```python
# create the dataset instance from the image folder
dataset = CustomDataset(data_path)
# check how many images are available in the dataset
data_length = dataset.__len__()
data_length
```
