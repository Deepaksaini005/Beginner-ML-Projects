```python

import pandas as pd

from sklearn.cluster import AgglomerativeClustering
from sklearn.preprocessing import LabelEncoder

# Create a numeric dataset for hierarchical clustering
data = {
    "X": ["Cat", "Dog", "Kitechen set" , "Dr set" , "Bench", "Sofa"],

    "Y" : ["Animal", "Animal", "Toys", "Toys", "Furniture", "Furniture"]
}

df = pd.DataFrame(data)

# # Create Hierarchical Clustering model
# model = AgglomerativeClustering(
#     n_clusters=2,
#     metric="euclidean",
#     linkage="ward",
# )

# # Train the model and assign clusters
# clusters = model.fit_predict(df)

# # Add cluster labels to the dataset
# df["Cluster"] = clusters

# Display the result
print(df)
```

```python
le_x = LabelEncoder()
le_y = LabelEncoder()
df["X"] = le_x.fit_transform(df["X"])
df["Y"] = le_y.fit_transform(df["Y"])
```

```python
cluster_model = AgglomerativeClustering(n_clusters=3, linkage='ward', metric='euclidean')
df['Cluster'] = cluster_model.fit_predict(df[['X', 'Y']])
```
