```python
import pandas  as pd
import numpy   as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('titanic')
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.drop(['deck', 'embark_town', 'alive' ,'class','who'  , 'adult_male'], axis=1, inplace=True)
```

```python
df
```

```python
df.dropna(subset =['embarked'], inplace=True)
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['sex'] = le.fit_transform(df['sex'])
df['embarked'] = le.fit_transform(df['embarked'])
```

```python
df.isnull().sum()
```

```python
df['age'] = df['age'].fillna(df['age'].mean())

df = df.astype('int')
df
```

```python
x = df.drop('survived', axis=1)
y = df['survived']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
accuracy_score(y_test, y_pred)  # accuracy score
```

```python
confusion_matrix(y_test, y_pred)  # confusion matrix
```

```python
print(classification_report(y_test, y_pred))  # classification report
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
#KNN model
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train_scaled, y_train)  # fit the model to the training data
```

```python
y_pred_knn = knn.predict(x_test_scaled)
```

```python
accuracy_score(y_test, y_pred_knn)  # accuracy score for KNN model
```

```python
confusion_matrix(y_test, y_pred_knn)  # confusion matrix for KNN model
```

```python
print(classification_report(y_test, y_pred_knn))  # classification report for KNN model
```
