```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")

words = ["Python", "Dog", "Cat"]

embeddings = model.encode(words)

for word, embedding in zip(words, embeddings):
    print(word)
    print(embedding)
    print()
```
